```python
You have the data containing two record: "Quantity" and "Price per unit". 
How would you use areithmetic operators to calculate the total cost for each item, 
also calculate the total cost?
product 1, Quantity -5, price per unit -Rs. 250 
product 2, Quantity -25, price per unit -Rs. 656
product 3, Quantity -34, price per unit -Rs. 274
```

```python
product_1 = 5 * 250
product_2 = 25 * 656
product_3 = 34 * 274

print("The total price of product_1 is" , product_1)
print("The total price of product_2is" , product_2)
print("The total price of product_3 is" , product_3)

print("The total price is" , product_1 + product_2 + product_3)
```
