```python
# Array indexing and slicing
```

```python
# array[start:stop:step
# NOTE: stop-index is not included in the output

a = np.array([1,2,4,3,5,6,7])
print("Whole array is: ",a)
print("index value 2 is: ",a[2])
print("index value 2 to 4 is: ",a[1:4])
print("index value -2 is: ",a[-2])

print("slicing 2,3,6",a[1::2])
print("Slicing 1 3 7", a[::3])
print("Reverse", a[::-1])
print("First three element is:", a[:3])
print("Last three element is:", a[-3:])
 
```

```python
# Two dimentional array

arr2 = np.array([[1,2,3,4],[5,6,7,8]])
print(arr2)
```

```python
# indexing in two dimentional array
print("Two d array:\n",arr2)
print("Reverse:\n",arr2[::-1])
print("first row:\n",arr2[0])
print("last row:\n",arr2[1])
print("first row - third element:\n",arr2[0][2])
print("first row - first element:\n",arr2[0][0])
print("second row - first element:\n",arr2[1][0])
print("second row - first element:\n",arr2[1][0])

# Slicing in two d matrix

print("Slicing second row, 6 7 8 :",arr2[1][1:])
print("Slicing second row,5 6:",arr2[1][:2])

# slicing the row
print("Slicing:", arr2[:,3])

# Slicing [rows,cols]
print("Slicing",arr2[1:,3])
```

```python
# 6. Array stacking and splitting
```

```python
a = np.array([1,2,3,4,5])
b = np.array([6,7,8,9,10])

print("This is verticle stack\n",np.vstack((a,b))) # Verticle stacking - row wise

print("This is horizontal stack:\n",np.hstack((a,b))) # Horizontal stacking - cols wise
```

```python
c = np.array([[1,2,3,4],[5,6,7,8]])
print(c)
```

```python
print(np.vstack(c))
```

```python
print(np.hstack(c))
```

```python
print(np.hsplit(c,4))
```

```python
split = np.hsplit(c,4) # horizontal split
for s in split:
    print(s)
```

```python
vsplit = np.vsplit(c,2) # verticle split
for v in vsplit:
    print(v)
```

```python
# Mathematiclal Operations on Multiple Array
```

```python
ma = np.array([1,3,4,5,16])
mb = np.array([5,6,3,8,10])


print("ma:\n",ma)
print("mb:\n",mb)
print("ma * mb:\n",ma * mb)
print("ma + mb:\n",ma + mb)
print("Addition:\n",np.add(ma,mb)) # 
print("Subtraction:\n",np.subtract(ma,mb))
print("Multiplication:\n",np.multiply(ma,mb))
print("Division:\n",np.divide(mb,ma))
print("ma Divide by 3:\n",np.divide(ma,3))
print("ma divide by 2:\n",np.multiply(ma,2))
```

```python
a = np.array([1,2,3])
b = np.array([4,5,6])
print(np.dot(a,b)) # dot product 
# 1 * 4, 2 * 5, 3 * 6
# 4 + 10 + 18
```

```python
# Transpose

print(a.T)
```

```python
t =  np.array([[1,2,3],
               [4,5,6]])
print(t)
print()
print(t.T) # transpose the position of an array from row to col 
```
