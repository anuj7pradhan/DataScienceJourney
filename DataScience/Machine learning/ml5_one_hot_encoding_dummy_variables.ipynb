```python
# one hot encoding is the process of encoding the data into two different sets which fills with ones and zeros,
# eg. male changes to 1 and Female changes to 0
```

```python
import pandas as pd
import numpy as np
import seaborn as sns
```

```python
df = pd.read_csv("loan.csv")
```

```python
df.head()
```

```python
# Finding the nulls
df.isnull().sum()
```

```python
df["Gender"] = df["Gender"].fillna(df["Gender"].mode()[0])
```

```python
df
```

```python
df.isnull().sum()
```

```python
df["Married"] = df["Married"].fillna(df["Married"].mode()[0])
```

```python
df
```

```python
df.isnull().sum()
```

```python
# now ther is no any nulls in the Gender and Married columns
```

```python
# Puttoing in teh encoding variables
en_data = df[["Gender","Married"]]
en_data
```

```python
# Getting the encoding dummies data
pd.get_dummies(en_data).info()
```

```python
# The datas are in str format so now use the scikit learn for more
en_data.info()
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
#  Sparse matrix is mostly filled with zores and ones
ohe = OneHotEncoder(drop="first") # after drop first encoding, Gender_Female and Married_No aare dropped 
ar = ohe.fit_transform(en_data).toarray()
ar
```

```python
# pd.DataFrame(ar,columns=["Gender_Female","Gender_Male","Married_No","Married_Yes"])
pd.DataFrame(ar,columns=["Gender_Male","Married_Yes"])
# 1 represents male and 0 represents female on Gender_Male column
# 1 represents married and 0 represents unmarried on Married_yes column
```
