```python
# df.bfill() --> backward fill from the below
# df.bfill(axis = 1) backward fill with axis=1 fills the data from right column

# df.ffill() --> forward fill from the above (First row won't be filled)
# df.ffill(axis = 1) forward fill with axis=1 fills the data from left column

df = df.bfill()
```

```python
df
```

```python
# mode filling which helps us to fill the repeated data into the missing value

# Finding mode from the dataset
df["Gender"].mode()
```

```python
# df["Gender"].fillna(df["Gender"].mode()[0])
```

```python
# Filling mode value to missing values
df["Gender"] = df["Gender"].fillna(df["Gender"].mode()[0])
```
