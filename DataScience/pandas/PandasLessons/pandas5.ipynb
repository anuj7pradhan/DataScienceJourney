```python
# UPDATING THE DATAFRAME
people = {
    "firstName": ["Suncha", "sirish","sabin", "mika"],
    "lastName": ["voxx", "dalli", "rai", "dalli"],
    "email": ["Sunchavoxx@gmail.com", "sirishdalli@gmail.com","sabinrai@gmail.com", "mikadalli@gmail.com"]
}
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(people)
```

```python
df
```

```python
df.columns  # Executes the columns name
```

```python
df.columns = ["first_name", "last_name", "email"]   # changes the columns name 
```

```python
df
```

```python
# list comprehension method to change columns name
# columns name into upper case

df.columns = [x.upper() for x in df.columns]
df
```

```python
# list comprehension method to change it in lower case
df.columns = [x.lower() for x in df.columns] 
```

```python
df
```

```python
# Changing _ to space or vice-versa 
df.columns = df.columns.str.replace(" ","_")
df
```

```python
# rename the columns
df.rename(columns = {"first_name": "first", "last_name": "last"})
```

```python
df
```

```python
# inplace = True just changed it permanently
df.rename(columns = {"first_name": "first", "last_name": "last"}, inplace = True)
df
```
