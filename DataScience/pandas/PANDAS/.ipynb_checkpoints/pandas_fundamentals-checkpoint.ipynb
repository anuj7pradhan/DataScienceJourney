```python
# 4. Rows & Columns - Selections
```

```python
df[['Last Name']]
```

```python
df[['Name','Last Name']] # Selecting the columns
```

```python
# loc - Lable based index
# selects the rows
df.loc[df.Name =='Anil'] # Selecting Row in index based
```

```python
df.loc[0]
```

```python
df.loc[(df.Name == 'Anil') & (df.Salary >= 50000)] # Giving conditions
```

```python
# iloc - index-value based 

df.iloc[3]
```

```python
df.iloc[2:5] # [start:stop] excludes in iloc
```

```python
df.loc[0:4]  # [start:stop] includes in loc
```

```python
df['Age'] > 30 # Provides the bool value
```

```python
df[df['Age']>30] # Provide all the detailed data
```

```python
df['Salary'] >= 100000
```

```python
df[df['Salary'] >= 100000] # shows all the dataframe of the table
```

```python
df[(df['Salary'] >= 100000) & (df['Age'] <= 30 )] # df[(condition1) & (condition2)] 
# Salary that is greater than 100000 & Age < 30
```

```python
df.where(df['Salary'] >= 100000) # data that doesn't fits the condition is displayed as NaN
```

```python
df.where(df['Salary'] >= 100000, other = "Not Eligible") # other changed NaN to Not Eligible
```
