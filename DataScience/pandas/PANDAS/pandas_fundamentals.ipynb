```python
# Lesson 10. Concatination and Merge dataframes
```

```python
df1 = pd.DataFrame({
    'ID':[1,2,3],
    'Name':['A','B','C'],
    'Address':['Add1','Add2','Add3']
})
```

```python
df1
```

```python
df2 = pd.DataFrame({
    'ID':[1,2,3,4],
    'Score':[77,88,99,66]
})
```

```python
df2
```

```python
pd.concat([df1,df2],axis = 0) # vertical / row level / top on top
```

```python
pd.concat([df1,df2],axis = 1)  # horizontal / column level / side on side
```

```python
# merge = join
pd.merge(df1,df2, how = 'inner', on ='ID' )
```

```python
pd.merge(df1,df2, how = 'inner', left_on ='ID', right_on ='ID'  )
```
